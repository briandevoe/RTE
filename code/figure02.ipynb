```python
import numpy as np
import matplotlib.pyplot as plt

# Beam zenith angle (0–90 degrees)
theta_deg = np.linspace(0, 90, 500)
theta = np.deg2rad(theta_deg)

# ---------------------------------------------------
# G-functions
# ---------------------------------------------------

G_uniform = 0.5 * np.ones_like(theta)

G_planophile = 3*(1 + np.cos(theta)**2)/8

G_erectophile = 3*(2 + np.sin(theta)**2)/16

G_plagiophile = 5*(3 + np.cos(4*theta))/32

G_extremophile = 5*(3 - np.cos(4*theta))/28


# ---------------------------------------------------
# Plot
# ---------------------------------------------------

plt.figure(figsize=(9,6))

plt.plot(theta_deg, G_planophile, label="Planophile", linewidth=2)
plt.plot(theta_deg, G_erectophile, label="Erectophile", linewidth=2)
plt.plot(theta_deg, G_plagiophile, label="Plagiophile", linewidth=2)
plt.plot(theta_deg, G_extremophile, label="Extremophile", linewidth=2)

plt.plot(theta_deg, G_uniform, linestyle="--", label="Uniform")

plt.axhline(0.5, linestyle=":", color="gray")

plt.xlabel("Beam zenith angle θ (degrees)")
plt.ylabel("G(θ)")
plt.title("Analytical Forms of the Canopy G-Function")

plt.legend()
plt.grid(alpha=0.3)

plt.xlim(0,90)

plt.show()
```
