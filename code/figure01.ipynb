```python
import numpy as np
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# Leaf inclination angle (0° to 90°)
# ---------------------------------------------------------
theta_deg = np.linspace(0, 90, 500)
theta = np.deg2rad(theta_deg)

# ---------------------------------------------------------
# Leaf inclination distribution functions g_L(theta_L)
# ---------------------------------------------------------

# Uniform
g_uniform = np.ones_like(theta)

# Planophile (mostly horizontal leaves)
g_planophile = 3 * np.cos(theta)**2

# Erectophile (mostly vertical leaves)
g_erectophile = (3/2) * np.sin(theta)**2

# Plagiophile (mostly ~45 degrees)
g_plagiophile = (15/8) * np.sin(2*theta)**2

# Extremophile (horizontal + vertical peaks)
g_extremophile = (15/8) * np.cos(2*theta)**2

# Spherical distribution
g_spherical = (4/np.pi) * np.sin(theta)

# ---------------------------------------------------------
# Plot
# ---------------------------------------------------------

plt.figure(figsize=(9,6))

plt.plot(theta_deg, g_planophile, label="Planophile", linewidth=2)
plt.plot(theta_deg, g_erectophile, label="Erectophile", linewidth=2)
plt.plot(theta_deg, g_plagiophile, label="Plagiophile", linewidth=2)
plt.plot(theta_deg, g_extremophile, label="Extremophile", linewidth=2)
plt.plot(theta_deg, g_spherical, label="Spherical", linewidth=2)
plt.plot(theta_deg, g_uniform, label="Uniform", linestyle="--")

plt.xlabel("Leaf inclination angle θL (degrees)")
plt.ylabel("gL(θL)")
plt.title("Leaf Normal Inclination Distribution Functions")
plt.legend()
plt.grid(alpha=0.3)

plt.show()
```
